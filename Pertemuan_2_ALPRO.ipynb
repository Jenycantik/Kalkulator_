# Tipe Data

## Numeric

```python
a = 10
print(type(a))  # Output: <class 'int'>
print(a)
```

```python
b = 10.5
print(type(b))  # Output: <class 'float'>
print(b)
```

```python
c = 3 + 4j
print(type(c))  # Output: <class 'complex'>
print(c)
```

## String

```python
d = "Haloo!"
print(type(d))  # Output: <class 'str'>
print(d)
```

## Sequence

```python
e = [1, 2, 3, 4]
print(type(e))  # Output: <class 'list'>
print(e)
```

```python
e.sort(reverse = True)
print(e)
```

```python
e.insert(9, 5)
print(e)
```

```python
f = (1, 2, 3, 4)
print(type(f))  # Output: <class 'tuple'>
print(f)
```

```python
g = range(5)
print(type(g))  # Output: <class 'range'>
print(g)
```

## Mapping

```python
h = {"name": "Falah", "age": 21}
print(type(h))  # Output: <class 'dict'>
print(h)
```

## Boolean

```python
i = True
print(type(i))  # Output: <class 'bool'>
print(i)
```

## Set

```python
j = {1, 2, 3, 4}
print(type(j))  # Output: <class 'set'>
print(j)
```

```python
k = frozenset([1, 2, 3, 4])
print(type(k))  # Output: <class 'frozenset'>
print(k)
```

```python
from typing import Final

PI: Final = 3.14159
GRAVITY: Final = 9.8

print("Nilai PI:", PI)
print("Nilai Gravitasi:", GRAVITY)
```
