DESKRIPSI PROGRAM: Program Deteksi Dini Anxiety Disorder adalah program yang membantu pengguna mengidentifikasi tanda-tanda awal gangguan kecemasan. Dengan memberikan gambaran tingkatan kecemasan dan mendorong konsultasi ke profesional jika diperlukan, seperti psikolog atau psikiater.






```python
#imort Modul
import sqlite3

#untuk membuat koneksi ke database SQLite
conn = sqlite3.connect("Deteksi_Dini_Anxiety_Disorder.db")
cursor = conn.cursor()

#fungsi untuk membuat database
#CREATE TABLE untuk membuat tabel baru jika belum ada
cursor.execute("""
            CREATE TABLE IF NOT EXISTS Deteksi_Dini_Anxiety_Disorder (
              id INTEGER PRIMARY KEY AUTOINCREMENT,
              nama TEXT,
              umur INTEGER,
              jenis_kelamin TEXT,
              score INTEGER
    )
""")
conn.commit()   #untuk menyimpan perubahan pada database

#fungsi untuk menyimpan data keDatabase
def insert_data(nama, umur, jenis_kelamin, score):
    cursor.execute('''
        INSERT INTO Deteksi_Dini_Anxiety_Disorder (nama, umur, jenis_kelamin, score)
        VALUES (?, ?, ?, ?)
    ''', (nama, umur, jenis_kelamin, score))
    conn.commit()
    print("Data berhasil disimpan")     #ditampilkan jika data berhasil disimpan


#fungsi untuk menampilkan data
def read_data():
    cursor.execute("SELECT * FROM Deteksi_Dini_Anxiety_Disorder")
    rows = cursor.fetchall()      #untuk mengambil semua hasil quary

    if rows:        #untuk memeriksa apakah data ada di database
        print("\n======== Data Pasien ========")
        for row in rows:      #untuk menampilkan setiap baris data
            print(f"ID: {row[0]}, Nama: {row[1]}, Umur: {row[2]}, Jenis Kelamin: {row[3]}, Score: {row[4]}")
        print("=========================")
    else:
        print("\nTidak ada data yang tersedia.\n")    #ditampilkan jika data tidak ada

#fungsi untuk update data
def update_data(user_id, nama, umur, jenis_kelamin):
    cursor.execute("SELECT * FROM Deteksi_Dini_Anxiety_Disorder WHERE id = ?", (user_id,))    #untuk cek data dengan id
    if cursor.fetchone():   #jika id ditemukan
      if jenis_kelamin:
        cursor.execute("""
            UPDATE Deteksi_Dini_Anxiety_Disorder
            SET nama = ?, umur = ?, jenis_kelamin = ?
            WHERE id = ?
        """, (nama, umur, jenis_kelamin, user_id))
      else:
        cursor.execute("""
            UPDATE Deteksi_Dini_Anxiety_Disorder
            SET nama = ?, umur = ?
            WHERE id = ?
        """, (nama, umur, user_id))
      conn.commit()       #untuk menyimpan perubahan
      print("Data berhasil diupdate")     #jika berhasil di update
    else:
      print("Data dengan ID tersebut tidak ditemukan.")     #jika id tidak ditemukan

#fungsi untuk menghapus data
def delete_data(user_id):
    cursor.execute("DELETE FROM Deteksi_Dini_Anxiety_Disorder WHERE id = ?", (user_id,))
    conn.commit()  # Menyimpan perubahan pada database
    if cursor.rowcount > 0:
        print("Data berhasil dihapus")    #jika berhasil dihapus
        cursor.execute("VACUUM")         #untuk menyusun ulang tabel agar ID terurut kembali
        conn.commit()
    else:
        print("Data dengan ID tersebut tidak ditemukan.")   #jika id tidak ditemukan

#fungsi untuk mencari data berdasarkan ID
def search_data():
    print("\nPencarian Data Pasien Berdasarkan ID")
    user_id = input("Masukkan ID pasien: ").strip()
    cursor.execute("SELECT * FROM Deteksi_Dini_Anxiety_Disorder WHERE id = ?", (user_id,))
    row = cursor.fetchone()
    if row:
        print(f"\nData ditemukan:\nID: {row[0]}, Nama: {row[1]}, Umur: {row[2]}, Jenis Kelamin: {row[3]}, Score: {row[4]}")
    else:
        print("Data dengan ID tersebut tidak ditemukan.")

#fungsi untuk mengumpulkan data identitas pengguna
def identitas():
    nama = input("Masukkan nama Anda: ").strip()    # strip untuk memastikan input yang diberikan tidak mengandung spasi yang tidak penting
    while not nama:       #validasi input nama supaya tidak kosong
        nama = input("Nama tidak boleh kosong. Masukkan nama Anda: ").strip()

    umur = input("Masukkan umur Anda: ")
    while not umur.isdigit() or int(umur) <= 0:       #validasi input umur agar berupa angka positif
        umur = input("Umur harus angka positif. Masukkan umur Anda: ")

    jenis_kelamin = input("Masukkan jenis kelamin (Laki-Laki/Perempuan): ").strip().capitalize()    #untuk menjadikan kapital semua
    while jenis_kelamin not in ['Laki-laki', 'Perempuan']:        #validasi input jenis kelamin
        jenis_kelamin = input("Jenis kelamin harus 'Laki-Laki' atau 'Perempuan': ").strip().capitalize()

    print(f"\nNama: {nama}\nUmur: {umur}\nJenis Kelamin: {jenis_kelamin}")
    return nama, int(umur), jenis_kelamin

# Fungsi untuk menampilkan dan memproses kuesioner
def kuesioner():
    questions = [
        #tampilan daftar pertanyaan kuesioner
        "Apakah Anda merasa terancam ketika mendapatkan tekanan? (iya/tidak): ",
        "Apakah Anda merasa gugup pada saat melakukan hal baru? (iya/tidak): ",
        "Apakah Anda mudah kecewa ketika yang Anda lakukan tidak sesuai harapan? (iya/tidak): ",
        "Apakah Anda sering sesak nafas ketika menghadapi masalah? (iya/tidak): ",
        "Apakah Anda sering merasakan nyeri pada dada ketika sedang bersedih? (iya/tidak): ",
        "Apakah Anda mudah tersinggung? (iya/tidak): ",
        "Apakah Anda mudah bersedih? (iya/tidak): ",
        "Apakah Anda sulit menangkap informasi baru? (iya/tidak): ",
        "Apakah Anda cenderung menghindar ketika berada dalam situasi yang mengancam? (iya/tidak): ",
        "Apakah Anda akan mencari perlindungan ketika mendapatkan masalah? (iya/tidak): ",
        "Apakah Anda sering berbicara tidak tepat sesuai konsonan atau keseleo lidah? (iya/tidak): ",
        "Apakah Anda merasa mual ketika sedang menghadapi masalah? (iya/tidak): ",
        "Apakah Anda mudah pingsan ketika merasa takut? (iya/tidak): "
    ]
    # untuk menyimpan jawaban pengguna
    answers = []
    # untuk mengajukan pertanyaan satu per satu dan menerima jawaban
    for question in questions:
        answer = input(question)
        while answer.lower() not in ['iya', 'tidak']:    #validasi jawaban hanya 'iya' atau 'tidak'
            print("Jawaban tidak valid! Harap jawab dengan 'iya' atau 'tidak'.")
            answer = input(question)
        answers.append(answer.lower())  #untuk menyimpan jawaban dalam huruf kecil
    return answers

# Fungsi untuk menilai hasil tes kecemasan
def penilaian_hasil(answers):
    score = answers.count("iya")  #untuk  menghitung jumlah jawaban "Iya"
    print("\n======Hasil Tes Kecemasan Anda:=======")

    # Menampilkan hasil berdasarkan jumlah jawaban "Iya"
    if score <= 2:
        print("tidak ada kecenderungan kecemasan. Anda tampaknya tidak mengalami gejala kecemasan yang signifikan.")
    elif 3 <= score <= 4:
        print("Kecenderungan ringan. Gejala kecemasan dapat dikelola dengan teknik pengelolaan stres.")
    elif 5 <= score <= 7:
        print("Kecenderungan sedang. Gejala kecemasan lebih signifikan, perhatikan dan pertimbangkan untuk mencari dukungan.")
        print("Jangan ragu untuk mencari bantuan untuk mendukung kesehatan mental Anda.\n")
    else:
        print("Kecenderungan tinggi. Gejala kecemasan terdeteksi melebihi batas ambang wajar. Disarankan untuk berkonsultasi dengan profesional seperti psikolog atau psikiater.")
        print("Jangan ragu untuk mencari bantuan untuk mendukung kesehatan mental Anda.\n")

    # Menampilkan saran jika score kurang dari 4
    if score < 4:
        saran()   #ditampilkan jika score kurang dari 4

#fungsi untuk menjalankan menu
def main():
    pendahuluan()   #untuk menampilkan pendahuluan

    while True:
      print("\n===== Menu Deteksi Dini Anxiety Disorder =====")
      print("1. Mulai Tes Deteksi Dini Anxiety Disorder")
      print("2. Tampilkan Data")
      print("3. Update Data")
      print("4. Hapus Data")
      print("5. Cari Data")
      print("0. Keluar")
      print("\n=========================================")
      pilihan = input("Pilih opsi (1/2/3/4/5/0): ").strip()   #strip untuk memastikan input yang diberikan tidak mengandung spasi yang tidak penting

      if pilihan == '1':
        nama, umur, jenis_kelamin = identitas()
        answers = kuesioner()
        score = answers.count("iya")
        insert_data(nama, umur, jenis_kelamin, answers.count("iya"))
      elif pilihan == '2':
        read_data()
      elif pilihan == '3':
        user_id = int(input("Masukkan ID data yang ingin diupdate: "))
        nama = input("Masukkan nama baru: ").strip()
        umur = int(input("Masukkan umur baru: "))
        jenis_kelamin = input("Masukkan jenis kelamin baru (Laki-Laki/Perempuan): ").strip().capitalize() #strip untuk memastikan input yang diberikan tidak mengandung spasi yang tidak penting dan capatalize untuk menampilkan huruf kapital
        update_data(user_id, nama, umur, jenis_kelamin)

      elif pilihan == '4':
        user_id = int(input("Masukkan ID data yang ingin dihapus: "))
        delete_data(user_id)

      elif pilihan == '5':
        search_data()

      elif pilihan == '0':
        print("Terima kasih telah menggunakan program ini. jangan lupa tetap jaga kesehatan mental dan fisik anda. bahagia selalu✨")
        break       #untuk menghentikan program
      else:
        print("Pilihan tidak valid! Silakan pilih 1, 2, 3, 4, atau 0.")

# Fungsi untuk menampilkan pendahuluan tes Deteksi Dini Anxiety Disorder
def pendahuluan():
    print("\n========= Deteksi Dini Anxiety Disorder =========")
    print("Selamat datang di Deteksi Dini Anxiety Disorder!\n")
    print("Tujuan tes ini adalah untuk membantu Anda mengenali tanda-tanda kecemasan yang dapat memengaruhi kesehatan mental Anda.")
    print("PERINGATAN: Hasil tes ini hanya untuk indikasi awal, bukan diagnosis medis. Jika Anda merasa mengalami gejala yang lebih serius, disarankan untuk berkonsultasi dengan profesional.\n")


# Fungsi untuk memberikan saran pengelolaan Deteksi Dini Anxiety Disorder
def saran():
    print("\nPanduan Sederhana untuk Mengelola Kecemasan:")
    print("1. Pernafasan dalam (Deep Breathing)")
    print("2. Teknik relaksasi otot progresif")
    print("3. Berbicara dengan seseorang yang dipercaya")
    print("4. Aktivitas fisik seperti olahraga atau jalan kaki")
    print("5. Meditasi atau yoga")
    print("6. Makan makanan favorit")

#untuk mienjalankan program utama
if __name__ == '__main__':
  main()
```
